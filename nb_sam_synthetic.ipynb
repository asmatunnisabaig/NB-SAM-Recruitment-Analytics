```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import timedelta

fake = Faker('en_IN')
np.random.seed(42)
random.seed(42)

real_df = pd.read_excel('/content/NB_SAM_Recruitment_Cleaned.xlsx',
                        sheet_name='Recruitment_Data')

real_df = real_df.rename(columns={'Name': 'Candidate Name'})

COLUMNS = [
    'S.No.', 'Date', 'Day', 'Skill', 'Candidate Name', 'Phone Number',
    'Email', 'Referred By', 'Location',
    'Relevant Experience (Yrs)', 'Total Experience (Yrs)',
    'Notice Period (Days)', 'Current CTC', 'Expected CTC',
    'Profile Status', 'Client Name', 'Client Share Date',
    'Assigned Recruiter Follow-up', 'Client Database Duplicate',
    'Client Update', 'Client Remarks'
]


recruiters = ['Masarrat', 'Abhilasha', 'Sharf', 'Neha',
              'Abhilasha & Masarrat', 'Sharf & Neha']
recruiter_weights = [0.3247, 0.2435, 0.2435, 0.1494, 0.0130, 0.0130]
# null rate for Referred By = 5.19%
RECRUITER_NULL_RATE = 0.0519


skills = [
    'SAP MM', 'SENIOR ORACLE RETAIL CONSULTANT', 'SAP ABAP', 'SAP FICO',
    'DOT NET DEVELOPER', 'SAP BTP', 'FRONT END DEVELOPER',
    'AMAZON REQUIREMENT', 'AWS DEVOPS', 'SAM SPECIALIST', 'SAP SD',
    'TEST LEADS', 'JAVA FULL STACK', 'JAVA BACKEND WITH KAFKA',
    'SAP ESSP', 'SAP TM', 'AMAZON REQUIREMENT SOFT DEV ENG',
    'COLOCATION INFRASTRUCTURE LEAD', 'HCM ANALYST OR ADMINISTRATOR',
    'IT SERVICE NOW DEVELOPER', 'LDD', 'MDM INFORMATICA SAAS ENGINEER',
    'SALES FORCE DEVELOPER', 'SAP ATTP', 'SAP BASIS ARCH', 'SAP TM FUNC'
]
skill_weights = [
    0.1169, 0.1039, 0.0909, 0.0909, 0.0649, 0.0519, 0.0519,
    0.0390, 0.0390, 0.0390, 0.0390, 0.0390, 0.0260, 0.0260,
    0.0260, 0.0260, 0.0130, 0.0130, 0.0130, 0.0130, 0.0130,
    0.0130, 0.0130, 0.0130, 0.0130, 0.0130
]


statuses       = ['Shared', 'Internal Reject', 'Rejected by Client (Duplicate)']
status_weights = [0.7532, 0.2078, 0.0390]


clients = ['Sidra Force', 'Infosys', 'Westernacher']

locations = [None, 'Hyderabad', 'Bangalore', 'Pune', 'Chennai',
             'Mumbai', 'Kolkata', 'Bhubaneswar', 'Coimbatore', 'Delhi']
location_weights = [0.3247, 0.2338, 0.1688, 0.0779, 0.0649,
                    0.0390, 0.0260, 0.0260, 0.0260, 0.0130]


notice_periods = [
    0.0,
    30.0,
    60.0,
    10.0,
    16.0,
    20.0,
    23.0,
    40.0,
    90.0
]

notice_weights = [
    0.5833,
    0.1458,
    0.1458,
    0.0208,
    0.0208,
    0.0208,
    0.0208,
    0.0208,
    0.0208
]


skill_params = {
    'SAP MM':                           (13.2, 5.43, 13.71, 7.03, 46.23),
    'SAP FICO':                         (13.4, 4.36, 19.57, 2.88, 28.75),
    'SAP ABAP':                         (8.0,  2.76, 16.90, 7.44, 50.01),
    'SAP BTP':                          (5.7,  1.03,  8.93, 3.94, 64.73),
    'SAP SD':                           (13.2, 2.25, 20.00, 5.57, 15.52),
    'SAM SPECIALIST':                   (10.5, 2.70, 21.85,15.96, 36.25),
    'TEST LEADS':                       (11.9, 0.81, 18.67, 3.51, 31.64),
    'AWS DEVOPS':                       (10.3, 4.07, 11.90, 3.72, 67.74),
    'JAVA FULL STACK':                  (6.0,  2.05, 14.09, 2.37, 31.94),
    'JAVA BACKEND WITH KAFKA':          (5.4,  0.28,  5.77, 1.75,158.42),
    'IT SERVICE NOW DEVELOPER':         (9.8,  1.00, 28.00, 2.00, 20.00),
    'SALES FORCE DEVELOPER':            (12.0, 1.00, 20.00, 2.00, 15.00),
    'SAP TM FUNC':                      (7.0,  1.00, 20.00, 2.00, 15.00),
    'SAP BASIS ARCH':                   (7.1,  1.00, 18.90, 2.00, 15.00),
    'LDD':                              (6.2,  1.00, 17.00, 2.00, 15.00),
    'MDM INFORMATICA SAAS ENGINEER':    (10.5, 1.00, 13.00, 2.00, 20.00),
    'SAP ATTP':                         (7.8,  1.00,  7.60, 1.00, 15.00),
    'SENIOR ORACLE RETAIL CONSULTANT':  (9.0,  3.00, 14.50, 4.00, 25.00),
    'DOT NET DEVELOPER':                (5.0,  2.00, 10.00, 3.00, 30.00),
    'FRONT END DEVELOPER':              (4.0,  1.50,  8.00, 2.00, 35.00),
    'AMAZON REQUIREMENT':               (5.0,  2.00,  9.00, 2.50, 30.00),
    'SAP ESSP':                         (8.0,  2.00, 15.00, 3.00, 25.00),
    'SAP TM':                           (7.0,  2.00, 16.00, 3.00, 20.00),
    'AMAZON REQUIREMENT SOFT DEV ENG':  (2.5,  0.50,  6.40, 1.00, 30.00),
    'COLOCATION INFRASTRUCTURE LEAD':   (10.0, 2.00, 18.00, 3.00, 20.00),
    'HCM ANALYST OR ADMINISTRATOR':     (6.0,  2.00, 12.00, 2.50, 25.00),
}


monthly_volumes = {

    '2025-12': 100, '2026-01': 130, '2026-02': 100,

    '2026-03': 80,  '2026-04': 70,  '2026-05': 80,

    '2026-06': 70,  '2026-07': 93

}


day_weights = {
    0: 0.0649, 1: 0.2727, 2: 0.1818,
    3: 0.2597, 4: 0.0649, 5: 0.1558, 6: 0.0000
}


null_rates = {
    'Expected CTC':                  0.4286,
    'Email':                         0.3896,
    'Client Share Date':             0.3766,
    'Notice Period (Days)':          0.3766,
    'Relevant Experience (Yrs)':     0.3766,
    'Total Experience (Yrs)':        0.3766,
    'Current CTC':                   0.3636,
    'Location':                      0.3247,
    'Phone Number':                  0.3247,
    'Assigned Recruiter Follow-up':  0.0909,
}

def null(col):
    return random.random() < null_rates.get(col, 0.0)

# Exact Assigned Recruiter Follow-up values from real data
followup_values = ['Masarrat', 'Sharf', 'Abhilasha', 'Neha',
                   'Sharf & Neha']
followup_weights = [0.35, 0.25, 0.25, 0.10, 0.05]

# Exact Client Update & Remarks from real data
def get_outcome(status):
    if status == 'Internal Reject':
        return ('Not Selected (Internal Reject)',
                'Not Selected (Internal Reject)')
    if status == 'Rejected by Client (Duplicate)':
        return ('Not Selected (Duplicate)',
                'Not Selected (Duplicate)')
    # Shared — match real data outcomes
    outcome = random.choices(
        ['Not Selected', 'Reached Interview Phase', None],
        weights=[0.70, 0.10, 0.20]
    )[0]
    if outcome == 'Not Selected':
        return ('Not Selected', 'Not Selected')
    elif outcome == 'Reached Interview Phase':
        return ('Reached Interview Phase',
                'Certificated Not Provided by Candidate')
    else:
        return (None, None)



def get_random_date(year_month_str):
    import calendar
    year, month = map(int, year_month_str.split('-'))
    _, days_in_month = calendar.monthrange(year, month)
    valid_days = [d for d in range(1, days_in_month + 1)
                  if pd.Timestamp(year, month, d).weekday() != 6]
    dow_for_day = [pd.Timestamp(year, month, d).weekday()
                   for d in valid_days]
    w = [day_weights[dow] for dow in dow_for_day]
    total = sum(w)
    w = [x / total for x in w]
    day = random.choices(valid_days, weights=w)[0]
    return pd.Timestamp(year, month, day)

def generate_experience(skill):
    p = skill_params.get(skill, (7.0, 2.0, 12.0, 3.0, 25.0))

    # Generate total experience
    total = round(
        min(15.0, max(1.0, np.random.normal(p[0], p[1]))),
        1
    )

    # Relevant experience is usually 70–95% of total experience
    relevant = round(
        min(total, max(0.5, total * random.uniform(0.70, 0.95))),
        1
    )

    return relevant, total

def generate_ctc(skill):
    p = skill_params.get(skill, (7.0, 2.0, 12.0, 3.0, 25.0))
    current = round(min(40.0, max(3.0, np.random.normal(p[2], p[3]))), 2)
    hike = max(5.0, np.random.normal(p[4], p[4] * 0.2))
    expected = round(min(45.0, current * (1 + hike / 100)), 2)
    return current, expected

def get_client(status, month_str):
    if status != 'Shared':
        return ('N/A (Internal Reject)' if status == 'Internal Reject'
                else 'N/A (Duplicate)')
    client_list = ['Sidra Force', 'Infosys', 'Westernacher']
    if month_str in ['2025-12', '2026-01']:
        w = [0.50, 0.30, 0.20]
    elif month_str in ['2026-02', '2026-03']:
        w = [0.68, 0.22, 0.10]
    else:
        w = [0.62, 0.28, 0.10]
    return random.choices(client_list, weights=w)[0]


rows = []
row_id = len(real_df) + 1

for month_str, count in monthly_volumes.items():
    for _ in range(count):
        skill    = random.choices(skills, weights=skill_weights)[0]
        status   = random.choices(statuses, weights=status_weights)[0]
        date     = get_random_date(month_str)
        client   = get_client(status, month_str)
        outcome, remark = get_outcome(status)

        # Recruiter
        if random.random() < RECRUITER_NULL_RATE:
            recruiter = None
        else:
            recruiter = random.choices(recruiters, weights=recruiter_weights)[0]

        # Experience
        rel_exp, tot_exp = generate_experience(skill)
        if null('Relevant Experience (Yrs)'): rel_exp = None
        if null('Total Experience (Yrs)'):    tot_exp = None

        # CTC
        if null('Current CTC'):
            current_ctc = None
            expected_ctc = None if null('Expected CTC') else generate_ctc(skill)[1]
        else:
             current_ctc, expected_ctc = generate_ctc(skill)
             if null('Expected CTC'): expected_ctc = None

        # Notice period
        notice = (None if null('Notice Period (Days)')
                  else random.choices(notice_periods, weights=notice_weights)[0])

        # Client Share Date
        if status == 'Shared' and not null('Client Share Date'):
            share_date = date + timedelta(days=random.randint(2, 7))
        else:
            share_date = None

        # Assigned Recruiter Follow-up
        if null('Assigned Recruiter Follow-up'):
            followup = None
        else:
            followup = random.choices(followup_values,
                                      weights=followup_weights)[0]

        rows.append({
            'S.No.'                      : row_id,
            'Date'                       : date,
            'Day'                        : date.strftime('%A'),
            'Skill'                      : skill,
            'Candidate Name'             : fake.name(),
            'Phone Number'               : (None if null('Phone Number')
                                            else fake.numerify('##########')),
            'Email'                      : (None if null('Email')
                                            else fake.email()),
            'Referred By'                : recruiter,
            'Location'                   : random.choices(
                                            locations,
                                            weights=location_weights)[0],
            'Relevant Experience (Yrs)'  : rel_exp,
            'Total Experience (Yrs)'     : tot_exp,
            'Notice Period (Days)'       : notice,
            'Current CTC'               : current_ctc,
            'Expected CTC'              : expected_ctc,
            'Profile Status'            : status,
            'Client Name'               : client,
            'Client Share Date'         : share_date,
            'Assigned Recruiter Follow-up': followup,
            'Client Database Duplicate' : ('Yes' if status ==
                                           'Rejected by Client (Duplicate)'
                                           else 'No'),
            'Client Update'             : outcome,
            'Client Remarks'            : remark,
        })
        row_id += 1

synthetic_df = pd.DataFrame(rows, columns=COLUMNS)


master_df = pd.concat([real_df[COLUMNS], synthetic_df],
                       ignore_index=True)
master_df['S.No.'] = range(1, len(master_df) + 1)


synthetic_df.to_csv('NB_SAM_Recruitment_Synthetic.csv', index=False)
master_df.to_csv('NB_SAM_Recruitment_Master.csv', index=False)
master_df.to_excel('NB_SAM_Recruitment_Master.xlsx', index=False)


print("=" * 55)
print("  DATASET SUMMARY")
print("=" * 55)
print(f"  Real rows       : {len(real_df)}")
print(f"  Synthetic rows  : {len(synthetic_df)}")
print(f"  Master total    : {len(master_df)}")
print(f"\n  Profile Status split (synthetic):")
print(synthetic_df['Profile Status'].value_counts())
print(f"\n  Top 5 Skills (synthetic):")
print(synthetic_df['Skill'].value_counts().head())
print(f"\n  Recruiter split (synthetic):")
print(synthetic_df['Referred By'].value_counts(dropna=False))
print(f"\n  Client split — Shared only (synthetic):")
shared = synthetic_df[synthetic_df['Profile Status'] == 'Shared']
print(shared['Client Name'].value_counts())
print(f"\n  Columns: {list(master_df.columns)}")
print("=" * 55)
```
